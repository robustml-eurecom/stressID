# Simple rule labels
#### Relax, valence and arousal

```python
#path_data = "Dataset/Physiological/"
path_data = "/Users/havachaptoukaev/Desktop/PhD/All projects/2 - Stress Wearables/StressID Dataset/Physiological/"
path_save = ""
```

```python
filelist= [f for f_sub in [f_names for root,d_names,f_names in os.walk(path_data)] for f in f_sub]
print("Should be :", 65*12 , ", there is:", len(filelist))
names = []
labs_binary = []
labs = []
```

```python
filelist.sort()
```

```python
lab_relax = []
lab_valence = []
lab_arousal = []
```

```python
#### RELAX
for f in filelist:
     if not f.startswith('.'):
        split = f.split('_')
        subject = split[0]
        task = split[1].split('.')[0]
        
        if not task == 'Baseline':
            task_stress = task+'_relax'
            ans = labels[subject].loc[labels['Tasks'] == task_stress]
            names.append(subject+'_'+task)
            if not ans.isnull().all():
                #print('--')
                ans = int(ans)
                l_b =  0 if ans < 6 else 1  #[0-5][6-10]
                lab_relax.append(l_b)
        #print("------------")
```

```python
#### VALENCE
for f in filelist:
     if not f.startswith('.'):
        split = f.split('_')
        subject = split[0]
        task = split[1].split('.')[0]
        
        if not task == 'Baseline':
            task_stress = task+'_valence'
            ans = labels[subject].loc[labels['Tasks'] == task_stress]
            
            if not ans.isnull().all():
                #print('--')
                ans = int(ans)
                l_b =  0 if ans < 6 else 1  #[0-5][6-10]
                lab_valence.append(l_b)
                
            else:
                lab_valence.append(np.nan)
        #print("------------")
```

```python
#### AROUSAL
for f in filelist:
     if not f.startswith('.'):
        split = f.split('_')
        subject = split[0]
        task = split[1].split('.')[0]
        
        if not task == 'Baseline':
            task_stress = task+'_arousal'
            ans = labels[subject].loc[labels['Tasks'] == task_stress]
            
            if not ans.isnull().all():
                #print('--')
                ans = int(ans)
                l_b =  0 if ans < 6 else 1  #[0-5][6-10]
                lab_arousal.append(l_b)
                
            else:
                lab_arousal.append(np.nan)
        #print("------------")
```

### Export

```python
df = {'subject/task':names, 'binary-relax': lab_relax, 'binary-valence': lab_valence, 'binary-arousal': lab_arousal}
labels_export = pd.DataFrame(df)
labels_export
```
